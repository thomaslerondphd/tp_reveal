## 2.

```python
import csv
def lire_csv(chemin):
    with open(chemin, mode='r', encoding='utf-8') as fichier:
        lecteur = csv.DictReader(fichier)
        return list(lecteur)
```

## 3.

```python
def compter_par(lignes, colonnes):
    compteur = {}
    for l in lignes:
        for c in colonnes:
            valeur = l.get(c)
            compteur[valeur] = compteur.get(valeur, 0) + 1
    return compteur
```

## 4.

```python
def total_par(lignes, colonne_groupe, colonne_valeur):
    totaux = {}
    for l in lignes:
        groupe = l.get(colonne_groupe)      
        try:
            valeur = int(l.get(colonne_valeur, 0))
        except (ValueError, TypeError):
            valeur = 0
            
        totaux[groupe] = totaux.get(groupe, 0) + valeur
        
    return totaux
```

## 5.

```python
class Site:
    def __init__(self, 
                code_site, 
                nom, 
                commune, 
                annee_ouverture, 
                surface_m2):
        
        self.code_site = code_site
        self.nom = nom
        self.commune = commune
        self.annee_ouverture = int(annee_ouverture)
        self.surface_m2 = int(surface_m2)

    def anciennete(self, 
                   annee_reference):
        return annee_reference - self.annee_ouverture

    def __str__(self):
        return f"{self.code_site} - {self.nom} ({self.commune}), {self.surface_m2} m2"
```

## 6.

```python
def reservations_en_cours(reservations):
    en_cours = []
    for res in reservations:
        date_retour = res.get("date_retour_reelle", "")
        if not date_retour :
            en_cours.append(res)
    return en_cours
```
