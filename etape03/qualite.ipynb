## 4.
1. Sur materiel.csv

    - Doublons stricts (7 lignes) : Suppression de la ligne. Ce sont des lignes en double parfait issues d'une erreur d'export ou de saisie, il faut les supprimer pour éviter de fausser les comptages de stock.

    - Année_achat vide (9 lignes) : Conservation avec signalement (valeur manquante). L'année d'achat est inconnue dans le système source ; on la laisse vide conformément à la règle de non-substitution.

    - Année_achat postérieure à 2026 (5 lignes) : Mise à valeur manquante (NaN). Ces années dans le futur (ex: 2031) sont des erreurs manifestes de saisie ; on les retire pour ne pas fausser l'âge du matériel, sans inventer d'année.

    - Valeur_achat vide (11 lignes) : Conservation avec signalement (valeur manquante). Le coût d'achat n'est pas renseigné pour ces équipements ; la valeur est laissée vide sans être remplacée par zéro.

    - Catégorie écrite intégralement en majuscules (6 lignes) : Correction automatique. Il s'agit d'un problème d'harmonisation textuelle (ex: "SONORISATION" au lieu de "Sonorisation") que l'on normalise pour les futurs regroupements.

    - Catégorie avec espaces parasites (5 lignes) : Correction automatique. Des espaces superflus en début ou fin de chaîne sont nettoyés (via .strip()) pour éviter de créer des catégories fictives.

    - Valeur_achat écrite avec une virgule décimale (5 lignes) : Correction automatique. Le format français avec virgule (ex: 180,08) est converti en format standard avec point (180.08) pour permettre les calculs numériques.

2. Sur associations.csv

    - Code_postal vide (9 lignes) : Conservation avec signalement (valeur manquante). L'information géographique est absente pour ces structures, on conserve la valeur vide sans l'inventer.

    - Type_association valant 'inconnu' (7 lignes) : Conservation avec signalement. Le type n'a pas pu être qualifié lors de l'enregistrement ; la mention est conservée en l'état pour analyse ultérieure.

    - Date_convention au format JJ/MM/AAAA (6 lignes) : Correction automatique. Ces dates sont converties au format standard international AAAA-MM-JJ pour garantir l'homogénéité temporelle de la table.

3. Sur reservations.csv

    - Doublons stricts (18 lignes) : Suppression de la ligne. Des réservations ont été dupliquées par erreur dans le fichier journal, il convient de les purger.

    - id_materiel ne correspondant à aucun matériel (21 lignes) : Suppression de la ligne (ou mise à valeur manquante). Ces réservations pointent vers des identités de matériel fantômes (9999), rendant le prêt incohérent lors des jointures.

    - Date_retour_reelle antérieure à date_debut (15 lignes) : Mise à valeur manquante (ou suppression). Une date de retour effective antérieure à la date de début de prêt est une aberration chronologique de saisie.

    - Quantité égale à zéro (6 lignes) : Suppression de la ligne (ou correction). Réserver une quantité nulle n'a pas de sens métier, ces lignes de réservation sont invalides.

    - /!\Cas métier normal — Prêts en cours (72 lignes) : Conservation

```python
import os

os.makedirs("sorties", exist_ok=True)

df_materiel = df_materiel.drop_duplicates().copy()

df_materiel['categorie'] = df_materiel['categorie'].astype(str).str.strip()

mapping_categories = {
    'SONORISATION': 'Sonorisation',
    'Sonorisation': 'Sonorisation',
    'Mobilier': 'Mobilier',
    'Eclairage': 'Eclairage',
    'Cuisine': 'Cuisine',
    'Tente et barnum': 'Tente et barnum',
    'SECURITE': 'Securite',
    'Securite': 'Securite'
}
df_materiel['categorie'] = df_materiel['categorie'].map(mapping_categories).fillna(df_materiel['categorie'])
df_materiel['valeur_achat'] = (
    df_materiel['valeur_achat']
    .astype(str)
    .str.replace(',', '.', regex=False)
)
df_materiel['valeur_achat'] = pd.to_numeric(df_materiel['valeur_achat'], errors='coerce')

df_materiel['annee_achat'] = pd.to_numeric(df_materiel['annee_achat'], errors='coerce')
df_materiel.loc[df_materiel['annee_achat'] > ANNEE_COURANTE, 'annee_achat'] = np.nan

df_materiel.to_csv("sorties/materiel_propre.csv", index=False, encoding="utf-8")

df_associations = df_associations.drop_duplicates().copy()

df_associations['date_convention'] = pd.to_datetime(
    df_associations['date_convention'], 
    format='mixed', 
    errors='coerce'
).dt.strftime('%Y-%m-%d')

df_associations.to_csv("sorties/associations_propre.csv", index=False, encoding="utf-8")

df_reservations = df_reservations.drop_duplicates().copy()
df_reservations = df_reservations[df_reservations['id_materiel'].isin(df_materiel['id_materiel'])].copy()

df_reservations['quantite'] = pd.to_numeric(df_reservations['quantite'], errors='coerce')
df_reservations = df_reservations[df_reservations['quantite'] > 0].copy()

condition_chronologie_valide = (
    df_reservations['date_retour_reelle'].isna() | 
    (df_reservations['date_retour_reelle'] >= df_reservations['date_debut'])
)
df_reservations = df_reservations[condition_chronologie_valide].copy()

for col in ['date_debut', 'date_fin_prevue', 'date_retour_reelle']:
    df_reservations[col] = df_reservations[col].dt.strftime('%Y-%m-%d').fillna('')

df_reservations.to_csv("sorties/reservations_propre.csv", index=False, encoding="utf-8")
```
